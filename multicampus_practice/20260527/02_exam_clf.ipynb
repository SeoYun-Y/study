```python
import requests
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
import platform
```

```python
url = 'https://archive-api.open-meteo.com/v1/archive'

params = {
    'latitude' : 37.5665,
    'longitude' : 126.9780,
    'start_date' : '2025-01-01',
    'end_date' : '2025-12-31',
    'hourly' : 'temperature_2m,relative_humidity_2m,surface_pressure,wind_speed_10m,rain'
}

res = requests.get(url, params = params)
data = res.json()
data
```

```python
# 날씨 데이터들을 따로 추출
df = pd.DataFrame(data['hourly'])
df.head(1)
```

```python
df.columns = ['측정시간','기온','상대습도','지면기압','풍속','강수량']
df.head(10)
```

```python
df.info()
```

#### 연습 문제
1. 날씨 데이터에서 분류를 할 데이터가 존재하지 않는다
    - 강수량 데이터를 기준으로 0보다 큰 경우라면 1(비), 아니라면 0(비x) 데이터를 이용하여 target 컬럼을 생성
    - target 데이터의 데이터의 분포 확인
    - 학습 데이터 ['기온','상대습도','지면기압','풍속']
2. Dataset을 만들어서 구간 3일치 데이터(72)로 잡는다
3. DataLoader를 생성 batch = 64로 생성, shuffle = False
4. 측정시간 컬럼을 시계열 데이터로 변경
5. LSTM 모델을 정의
    - LSTM은 기본값들을 사용 batch_first = True로 설정
6. 모델 생성, 손실함수, 옵티마이저 구성
7. 반복 학습( 학습의 횟수는 20 )
    - loss 값 반복마다 출력
    - 정확도를 출력(후 순위) - 한번의 학습마다 평균 정확도
8. 학습된 모델을 이용하여 예측

```python
# 파생변수 target 생성
# (df['강수량'] > 0).astype(int)
df['target'] = df['강수량'].map(
    lambda x : 1 if x > 0 else 0
)
```

```python
df['target'].value_counts()
```

```python
x = df[['기온','상대습도','지면기압','풍속']].values
y = df['target'].values
```

```python
# x의 데이터에서 MinMaxScaler를 이용하여 데이터 스케일링
x_sc = MinMaxScaler().fit_transform(x)
```

```python
class windowDS(Dataset):
    def __init__(self, _x, _y, _window):
        self.x = _x
        self.y = _y
        self.window = _window
        self.n = len(_x) - _window

    def __len__(self):
        return self.n
    
    def __getitem__(self, idx):
        x = self.x[idx:idx+self.window]
        y = self.y[idx+self.window]

        # 독립 변수의 데이터는 실수형태로 전환
        x_tensor = torch.tensor(x).float()
        # 종속 변수는 분류 모델을 이용하기 때문에 정수 형태로 전환
        y_tensor = torch.tensor(y).long()

        return x_tensor, y_tensor

# 72 : 72시간 의미(3일)
window = 72
train_ds = windowDS(x_sc, y, window)

# shuffle -> False , 균등하게 데이터를 확인하고 학습하기 위함
train_dl = DataLoader(train_ds, batch_size = 64, shuffle=False)
```

```python
# 분류형 LSTM 모델을 정의
class LSTMclf(nn.Module):
    def __init__(
            self,
            input_size,
            hidden_size
    ):
        super(LSTMclf, self).__init__()
        self.lstm = nn.LSTM(
            input_size = input_size,
            hidden_size = hidden_size,
            batch_first = True
        )
        # 분류 모델에서는 마지막 층의 아웃풋 피쳐의 개수는 분류 개수와 동일하게 맞춰준다
        self.model = nn.Linear(hidden_size, 2)
    
    def forward(self, x):
        out, (h_n, c_n) = self.lstm(x)
        # h_n[-1], c_n[-1] : bidirectional이 False인 경우에는 사용 가능 -> True인 경우에는 역방향의 아웃풋 피쳐만 불러온다
        # out : 3차원 데이터셋 -> [batch, window, feature]
        # 2차원 데이터셋의 형태로 변경 -> out[ :, -1, : ] -> [batch, feature]
        pred = self.model(out[:, -1, :])
        return pred
```

```python
model = LSTMclf(input_size = x.shape[1], hidden_size=32)
model
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr = 0.001)
```

```python
from sklearn.metrics import accuracy_score
import numpy as np
```

```python
# 반복 학습(loss, acc)
model.train()

for epoch in range(20):
    total_loss, total_n = 0.0, 0
    current1 = current2 = 0
    acc_list = []

    for x, y in train_dl:
        # 피쳐가 2개인 데이터 (예측값)
        res = model(x)
        loss = criterion(res, y)
        optimizer.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss += loss.item()
        total_n += y.size(0)
        _, pred = torch.max(res, 1)
        # 분류 (예측값을 이용해 높은 부분의 인덱스를 출력)
        acc = accuracy_score(y.numpy(), pred.numpy())
        acc_list.append(acc)
        current1 += (y == pred).sum().item()
        

    train_loss = total_loss / len(train_dl)
    mean_current = (current1 / max(total_n, 1)) * 100
    current2 = np.mean(acc_list)
        
    print(f'Epoch : {epoch+1}, EntropyLoss : {round(total_loss, 3)}, Acc1 : {round(mean_current, 3)}, Acc2 : {round(current2, 3)}')
```

```python
# 예측
model.eval()

eval_loader = DataLoader(train_ds, batch_size = 1000, shuffle = False)

# 첫번째 데이터만 로드
x, y = next(iter(eval_loader))

with torch.no_grad():
    test_out = model(x)
    _, test_pred = torch.max(test_out,1)
```
